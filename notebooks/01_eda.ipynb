```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(r'E:\Praxis\rossmann-sales\data\processed\clean_merged_data.csv', parse_dates=['Date'])
print("Average Sales when Open & Promo=1:", df[(df['Open']==1) & (df['Promo']==1)]['Sales'].mean())
print("Average Sales when Open & Promo=0:", df[(df['Open']==1) & (df['Promo']==0)]['Sales'].mean())
print("\nSales by StoreType (Open stores only):")
print(df[df['Open']==1].groupby('StoreType')['Sales'].mean())
```

```python
daily_sales = df.groupby('Date')['Sales'].mean()
plt.figure(figsize=(14, 5))
plt.plot(daily_sales.index, daily_sales.values)
plt.title('Average Daily Sales Over Time')
plt.xlabel('Date')
plt.ylabel('Sales')
plt.show()
```

```python
plt.figure(figsize=(10, 5))
sns.boxplot(x='StoreType', y='Sales', hue='Promo', data=df[df['Open'] == 1])
plt.title('Sales Distribution by Store Type and Promotion')
plt.show()
```
