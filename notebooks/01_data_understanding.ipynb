```python
import pandas as pd

train = pd.read_csv("../data/raw/train.csv")
store = pd.read_csv("../data/raw/store.csv")
```

##### 1. Dataset dimensions

```python
print("Train:", train.shape)
print("Store:", store.shape)
```

##### 2. Columns

```python
print(train.columns.tolist())
print(store.columns.tolist())
```

##### 3. Data types

```python
train.info()
store.info()
```

##### 4. Missing values

```python
print(train.isna().sum())
print("-" * 20)
print(store.isna().sum())
```

##### 5. Duplicate records

```python
print("Train duplicates:", train.duplicated().sum())
print("Store duplicates:", store.duplicated().sum())
```

##### 6. Basic statistics

```python
train.describe()
```

##### 7. Date range

```python
train["Date"] = pd.to_datetime(train["Date"])

print("Start:", train["Date"].min())
print("End:", train["Date"].max())
```

##### 8. Store count

```python
print("Unique stores:", train["Store"].nunique())
print("Store master records:", store["Store"].nunique())
```
