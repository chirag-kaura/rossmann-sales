## Step 13 — Baseline Model

A naive baseline was created using sales from the previous week
(`Sales_Lag_7`) as the prediction.

### Baseline Results

- Validation RMSPE: 0.4356
- Test RMSPE: 0.3843

The baseline provides a benchmark against which all machine learning models
will be evaluated.

Future models should demonstrate an improvement over this baseline while
maintaining good generalization on the unseen test period.

## Step 14 — First ML Model: Random Forest

A Random Forest regression model was trained using the engineered features.

### Results

- Validation RMSPE: 0.1578
- Baseline Validation RMSPE: 0.4356
- Improvement over baseline: approximately 63.8%

The Random Forest substantially outperformed the naive previous-week-sales
baseline.

This confirms that the engineered temporal, promotional, store-level, and
historical-sales features provide significant predictive value.

## Step 15 — Feature Importance

The Random Forest achieved a validation RMSPE of 0.1578.

The most important features were:

1. Open — 46.08%
2. Sales_Lag_14 — 30.56%
3. Sales_Rolling_Mean_7 — 6.59%
4. Sales_Lag_1 — 4.27%
5. Promo — 3.29%

The results show that store operating status and historical sales patterns
are the strongest predictors in the model.

The importance of lag and rolling features confirms that temporal information
is highly valuable for Rossmann sales forecasting.

The model also confirms the business importance of promotions identified
during EDA.

## Step 16 — XGBoost Model Comparison

XGBoost was trained using the same training and validation datasets and the
same engineered features used for the Random Forest model.

### Results

| Model | Validation RMSPE |
|---|---:|
| Naive Baseline | 0.4356 |
| Random Forest | 0.1578 |
| XGBoost | 0.1599 |

Random Forest achieved the lowest validation RMSPE and therefore remains the
best-performing model at this stage.

XGBoost also performed substantially better than the naive baseline, but did
not improve upon Random Forest.

The test dataset has not been used for model selection.

## Step 18 — Final Test Evaluation

The selected Random Forest model was evaluated on the completely unseen
July 2015 test dataset.

### Final Results

| Model | Test RMSPE |
|---|---:|
| Naive Baseline | 0.3843 |
| Random Forest | 0.1200 |

The final Random Forest achieved an RMSPE of 0.1200, representing a
68.78% improvement over the naive baseline.

The model was selected using the training and validation datasets, while
the test dataset was reserved for final evaluation.

Therefore, the test result provides an unbiased estimate of the final
model's forecasting performance.
